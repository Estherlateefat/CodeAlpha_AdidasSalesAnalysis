```python
Which Adidas product generates the highest total sales?
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Adidas_sales.csv")
```

```python
df.head()
```

```python
money_columns = ['Price per Unit', 'Total Sales', 'Operating Profit']

for col in money_columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
    )
```

```python
df['Units Sold'] = pd.to_numeric(df['Units Sold'], errors='coerce')
```

```python
df.dtypes
```

```python
product_sales = (
    df.groupby('Product')['Total Sales']
    .sum()
    .sort_values(ascending=True)
)

print(product_sales)
```

```python
plt.figure(figsize=(10, 6))

product_sales.plot(kind='barh')

plt.title('Total Sales by Product')
plt.xlabel('Total Sales ($)')
plt.ylabel('Product')

plt.tight_layout()

plt.savefig('total_sales_by_product.png', dpi=300, bbox_inches='tight')

plt.show()
```

### Insight

```python
Men's Street Footwear generated the highest total sales among the Adidas product categories, indicating strong customer demand for this product.
```

```python
Which region has the highest total sales?
```

```python

region_sales = (
df.groupby('Region')['Total Sales'].sum().sort_values(ascending=False)
)
plt.figure(figsize=(8, 5))
region_sales.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.savefig('total_sales_by_region.png', dpi=300, bbox_inches='tight')
plt.tight_layout()
plt.show()
```

### Insight:

```python
The West region recorded the highest total sales, suggesting that it is the strongest performing geographic market in the dataset
```

```python
df['Total Sales'] = (
    df['Total Sales']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
monthly_sales = (
    df.dropna(subset=['Invoice Date'])
      .groupby(
          df.dropna(subset=['Invoice Date'])['Invoice Date'].dt.to_period('M')
      )['Total Sales']
      .sum()
)

monthly_sales.index = monthly_sales.index.to_timestamp()

print(monthly_sales)
```

```python
df['Total Sales'].dtype
```

```python
plt.figure(figsize=(14, 7))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Adidas Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.savefig('monthly_adidas__sales_trend.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Insight:

```python
The monthly sales trend shows how Adidas sales fluctuate over time. The chart reveals periods of higher and lower sales, which may indicate seasonal patterns or changes in customer demand. 
The months with the highest sales can be investigated further to understand the factors contributing to the increase.
```
