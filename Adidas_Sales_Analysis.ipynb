```python
Q1 = df['Total Sales'].quantile(0.25)
Q3 = df['Total Sales'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Total Sales'] < lower_bound) |
    (df['Total Sales'] > upper_bound)
]

print("Number of potential Total Sales outliers:", len(outliers))
```

### Outlier Insight

The analysis identified unusually high Total Sales values that may be potential outliers. These values were not automatically removed because they may represent genuine high-value transactions and require further investigation.

### Which Adidas product generates the highest total sales?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Adidas_sales.csv")
```

```python
df.head()
```

```python
money_columns = ['Price per Unit', 'Total Sales', 'Operating Profit']

for col in money_columns:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False)
        .astype(float)
    )
```

```python
print("Number of duplicate records:", df.duplicated().sum())
```

```python
df.dtypes
```

```python
product_sales = (
    df.groupby('Product')['Total Sales']
    .sum()
    .sort_values(ascending=True)
)

print(product_sales)
```

```python
import os

os.makedirs("charts", exist_ok=True)
```

```python
plt.figure(figsize=(10, 6))

product_sales.plot(kind='barh')

plt.title('Total Sales by Product')
plt.xlabel('Total Sales ($)')
plt.ylabel('Product')

plt.tight_layout()
plt.savefig('charts/total_sales_by_product.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Insight

Men's Street Footwear generated the highest total sales among the Adidas product categories, indicating strong customer demand for this product.

Which region has the highest total sales?

```python

region_sales = (
df.groupby('Region')['Total Sales'].sum().sort_values(ascending=False)
)
plt.figure(figsize=(8, 5))
region_sales.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.savefig('charts/total_sales_by_region.png', dpi=300, bbox_inches='tight')

plt.tight_layout()
plt.show()
```

### Insight:

The West region recorded the highest total sales, suggesting that it is the strongest performing geographic market in the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Adidas_sales.csv")
```

```python
df['Total Sales'].dtype
```

```python
df['Invoice Date'] = pd.to_datetime(
    df['Invoice Date'],
    dayfirst=True,
    errors='coerce'
)

df['Total Sales'] = (
    df['Total Sales']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)

monthly_sales = (
    df.dropna(subset=['Invoice Date'])
      .groupby(
          df.dropna(subset=['Invoice Date'])['Invoice Date'].dt.to_period('M')
      )['Total Sales']
      .sum()
)

monthly_sales.index = monthly_sales.index.to_timestamp()

print(monthly_sales)
```

```python
plt.figure(figsize=(14, 7))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Adidas Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()

plt.savefig(
    'charts/monthly_adidas_sales_trend.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

### Insight:

The monthly sales trend shows how Adidas sales fluctuate over time. The chart reveals periods of higher and lower sales, which may indicate seasonal patterns or changes in customer demand. 
The months with the highest sales can be investigated further to understand the factors contributing to the increase.

### Sales by Product and Sales Method

```python
method_sales = (
    df.groupby('Sales Method')
      .agg({
          'Total Sales': 'sum',
          'Units Sold': 'sum'
      })
      .sort_values('Total Sales', ascending=False)
)

print(method_sales)
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Units Sold', y='Total Sales')

plt.title('Units Sold vs Total Sales')
plt.xlabel('Units Sold')
plt.ylabel('Total Sales ($)')
plt.tight_layout()
plt.savefig('charts/units_sold_vs_total_sales.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Insight

In-store sales generated the highest total sales revenue, while online sales recorded the highest number of units sold.
This shows that the sales method with the highest number of units does not necessarily generate the highest total sales.

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=df,
    x='Product',
    y='Total Sales',
    hue='Sales Method',
    estimator=sum
)

plt.title('Total Sales by Product and Sales Method')
plt.xlabel('Product')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('charts/total_sales_by_product_and_sales_method.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Insight:

Sales performance varies across products and sales methods. The visualization allows us to identify which
products perform strongly through In-store, Outlet, and Online channels. 
This can help Adidas determine which products should be prioritized in each sales channel.
